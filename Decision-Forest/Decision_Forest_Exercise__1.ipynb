```python
import kagglehub
import pandas as pd
import ydf
from sklearn.model_selection import train_test_split
```

```python
#@title Downloading the Dataset

# Download The wine quality dataset from kaggle
path = kagglehub.dataset_download("uciml/iris")

RAW_DATASET=pd.read_csv(path+ "/Iris.csv")

```

```python
#@title Preprocessing Data

seed=42

# removing the redundant column of 'ID'
df=RAW_DATASET.drop('Id',axis=1)
df

train_df, test_df = train_test_split(df, test_size=0.2, random_state=seed)

print(f"Train size: {len(train_df)}")
print(f"Test size: {len(test_df)}")
```

```python
#@title Training a Random Forest Model

model = ydf.RandomForestLearner(label="Species", random_seed=seed,
    categorical_algorithm="RANDOM",
    split_axis="SPARSE_OBLIQUE",
)

# Train the model (use .train() instead of .fit())
Trained_Model=model.train(train_df)

# Evaluate on training set
evaluation_Training = Trained_Model.evaluate(train_df)
print(f"______________________________________________________________________________________________________________________")
print(f"\nEvaluation Results of Model on the training set:\n\n")
print(evaluation_Training)

# describe function
print(f"______________________________________________________________________________________________________________________")
print(f"\nModel Describe Function:\n\n")
display(Trained_Model.describe())

# Evaluate on training set
evaluation_Test = Trained_Model.evaluate(test_df)
print(f"______________________________________________________________________________________________________________________")
print(f"\nEvaluation Results of Model on the TEST set:\n\n")
print(evaluation_Test)
```
